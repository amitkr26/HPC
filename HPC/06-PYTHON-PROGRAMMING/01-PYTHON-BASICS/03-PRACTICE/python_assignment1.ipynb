```
Born_Year = int(input("Enter the year in which you were born:"))
Current_Year = 2026
Age = Current_Year - Born_Year
print("Age is:", Age)
```

Write a script that asks the user how many cups of coffee they want to buy. If each cup costs $3.50, calculate their total cost. Print a message like: "Your total for 4 coffees is $14.0".

```
No_Cup = int(input("Enter how many cups you want to buy:"))
price = 3.50
Total_Cost = No_Cup * price
print("Your total for", No_Cup, "coffee is", Total_Cost)
```
